**2.9 IQR Outlier Count**

The Interquartile Range (IQR) method was used to identify potential outliers in each water-quality feature. A value is considered a potential outlier when it falls below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR. The number of potential outliers in each feature was counted for further assessment.

```python
outlier_counts = {}

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    outlier_counts[column] = len(outliers)

outlier_counts
```

## 4. Data Preprocessing
**Stage Owner: Araad HOF**

Preprocessing transforms the cleaned water data into an optimal representation for model training. 

In this stage, we implement:
1. An 80/20 stratified train/test split to maintain class proportions.
2. Feature scaling using `StandardScaler` to normalize feature magnitudes.
3. Leak-free processing using scikit-learn `Pipeline` to ensure all statistics are learned solely on the training partition.
4. Handling class imbalance using `class_weight='balanced'`.

### Rationale: Class Weights vs SMOTE
The dataset has 61% non-potable vs 39% potable samples. We utilize `class_weight='balanced'` instead of synthetic oversampling (SMOTE) because:
- It penalizes minority class errors directly in the objective function without introducing synthetic, interpolated samples that might blur complex chemical boundaries.
- It prevents data leakage risks and requires no external dependencies, keeping model comparisons completely fair.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve

RANDOM_STATE = 42
TARGET = 'Potability'

# Separate features and target
X = df.drop(columns=[TARGET])
y = df[TARGET]

# Stratified 80/20 train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE
)

print(f"Training Set: {X_train.shape[0]} samples (Potable: {y_train.mean():.1%})")
print(f"Testing Set:  {X_test.shape[0]} samples (Potable: {y_test.mean():.1%})")

```

### 4.2 Leak-Free Preprocessing Pipeline
We build a scikit-learn `Pipeline` encapsulating `StandardScaler`. This guarantees that feature scaling statistics are computed strictly on the training partition within each cross-validation fold, completely eliminating data leakage.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
def make_pipeline(model):
    """Encapsulates standard scaling and the classifier inside a leak-free pipeline."""
    return Pipeline([
        ('scaler', StandardScaler()),
        ('model', model)
    ])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scoring = ['accuracy', 'f1', 'roc_auc']

```

## 5. Modelling and Hyperparameter Tuning
**Stage Owners: Araad HOF (Baseline & Random Forest) & Ostino (Gradient Boosting & Deep Learning ANN)**

Per rubric requirements, we implement multiple distinct modeling approaches:
1. **Logistic Regression (Linear Baseline):** To benchmark linear separability.
2. **Random Forest (Tuned Ensemble):** Bagging ensemble tuned via 5-fold Stratified GridSearchCV.
3. **Gradient Boosting:** Sequential boosting ensemble.
4. **Artificial Neural Network (Deep Learning MLP):** Multi-layer perceptron with Batch Normalization, Dropout, and Sigmoid probability output.


### 5.1 Baseline Model: Logistic Regression
We train Logistic Regression with balanced class weights to evaluate linear performance.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
logreg_pipe = make_pipeline(
    LogisticRegression(class_weight='balanced', max_iter=1000, random_state=RANDOM_STATE)
)

lr_cv = cross_validate(logreg_pipe, X_train, y_train, cv=cv, scoring=scoring)
logreg_pipe.fit(X_train, y_train)

acc_mean = lr_cv['test_accuracy'].mean()
f1_mean = lr_cv['test_f1'].mean()
roc_mean = lr_cv['test_roc_auc'].mean()

print(f"Logistic Regression 5-Fold CV Accuracy: {acc_mean:.4f}")
print(f"Logistic Regression 5-Fold CV F1-Score: {f1_mean:.4f}")
print(f"Logistic Regression 5-Fold CV ROC-AUC:  {roc_mean:.4f}")

```

### 5.2 Tuned Model: Random Forest Classifier
We perform a grid search over key hyperparameters (`n_estimators`, `max_depth`, `min_samples_leaf`) using 5-fold stratified cross-validation on the training set, optimizing for F1-score on potable water.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
rf_pipe = make_pipeline(
    RandomForestClassifier(class_weight='balanced', random_state=RANDOM_STATE, n_jobs=-1)
)

param_grid = {
    'model__n_estimators': [100, 200],
    'model__max_depth': [10, 15],
    'model__min_samples_leaf': [1, 3, 5]
}

search = GridSearchCV(rf_pipe, param_grid, cv=cv, scoring=scoring, refit='f1', n_jobs=-1)
search.fit(X_train, y_train)
best_rf = search.best_estimator_

print(f"Best Random Forest Parameters: {search.best_params_}")
print(f"Best 5-Fold CV F1-Score: {search.best_score_:.4f}")

```

### 5.3 Advanced Ensemble: Gradient Boosting Classifier
Gradient Boosting builds sequential error-correcting decision trees to capture high-precision boundaries.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
gb_pipe = make_pipeline(
    GradientBoostingClassifier(n_estimators=150, learning_rate=0.05, max_depth=4, subsample=0.8, random_state=RANDOM_STATE)
)

gb_pipe.fit(X_train, y_train)
print('Gradient Boosting trained successfully.')

```

### 5.4 Deep Learning Architecture: Artificial Neural Network (Multilayer Perceptron - ANN)
**Stage Owner: Ostino**

To fulfill the bootcamp requirement for neural network architectures, we construct a deep Multilayer Perceptron:
- **Input Dimension:** 9 standardized features.
- **Hidden Layers:** Dense(64) -> BatchNorm -> Dropout(0.3) -> Dense(32) -> BatchNorm -> Dropout(0.2) -> Dense(16).
- **Activations:** ReLU in hidden layers to prevent vanishing gradients; Sigmoid in the output node to generate calibrated probabilities P(Potable = 1 | X).
- **Loss & Optimizer:** `binary_crossentropy` with `Adam(learning_rate=0.001)` and `EarlyStopping` on validation loss.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
    from tensorflow.keras.callbacks import EarlyStopping
    
    scaler_ann = StandardScaler()
    X_train_ann = scaler_ann.fit_transform(X_train)
    X_test_ann = scaler_ann.transform(X_test)
    
    tf.random.set_seed(RANDOM_STATE)
    ann_model = Sequential([
        Dense(64, activation='relu', input_shape=(X_train_ann.shape[1],)),
        BatchNormalization(),
        Dropout(0.3),
        
        Dense(32, activation='relu'),
        BatchNormalization(),
        Dropout(0.2),
        
        Dense(16, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    
    ann_model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    
    early_stop = EarlyStopping(monitor='val_loss', patience=12, restore_best_weights=True)
    
    history = ann_model.fit(
        X_train_ann, y_train,
        validation_split=0.2,
        epochs=80,
        batch_size=32,
        callbacks=[early_stop],
        verbose=0
    )
    
    ann_probs = ann_model.predict(X_test_ann).flatten()
    ann_preds = (ann_probs >= 0.5).astype(int)
    print('TensorFlow Keras ANN trained successfully.')
except Exception as e:
    from sklearn.neural_network import MLPClassifier
    scaler_ann = StandardScaler()
    X_train_ann = scaler_ann.fit_transform(X_train)
    X_test_ann = scaler_ann.transform(X_test)
    
    mlp = MLPClassifier(hidden_layer_sizes=(64, 32, 16), activation='relu', max_iter=300, random_state=RANDOM_STATE, early_stopping=True)
    mlp.fit(X_train_ann, y_train)
    ann_preds = mlp.predict(X_test_ann)
    ann_probs = mlp.predict_proba(X_test_ann)[:, 1]
    print('Sklearn MLP Neural Network trained successfully.')

```

## 6. Model Evaluation and Comparison
**Stage Owner: Ostino**

We evaluate all trained models on the held-out test set (656 samples). The evaluation includes:
1. Multi-metric summary table (Accuracy, Precision, Recall, F1, ROC-AUC).
2. Confusion matrices for all models.
3. ROC-AUC comparison curves.
4. Feature importance ranking.
5. Real-world public health domain interpretation of classification errors.


```python
# Assisted by AI, adapted for Group 10 water quality detection pipeline
lr_preds = logreg_pipe.predict(X_test)
lr_probs = logreg_pipe.predict_proba(X_test)[:, 1]

rf_preds = best_rf.predict(X_test)
rf_probs = best_rf.predict_proba(X_test)[:, 1]

gb_preds = gb_pipe.predict(X_test)
gb_probs = gb_pipe.predict_proba(X_test)[:, 1]

def evaluate_model(name, y_true, y_pred, y_prob):
    return {
        'Model': name,
        'Accuracy': round(accuracy_score(y_true, y_pred), 4),
        'Precision (Class 1)': round(precision_score(y_true, y_pred, pos_label=1), 4),
        'Recall (Class 1)': round(recall_score(y_true, y_pred, pos_label=1), 4),
        'F1-Score (Class 1)': round(f1_score(y_true, y_pred, pos_label=1), 4),
        'ROC-AUC': round(roc_auc_score(y_true, y_prob), 4)
    }

results = [
    evaluate_model('Logistic Regression', y_test, lr_preds, lr_probs),
    evaluate_model('Random Forest (Tuned)', y_test, rf_preds, rf_probs),
    evaluate_model('Gradient Boosting', y_test, gb_preds, gb_probs),
    evaluate_model('Neural Network (ANN)', y_test, ann_preds, ann_probs)
]

results_df = pd.DataFrame(results)
print('=== FINAL TEST SET EVALUATION BENCHMARK ===')
display(results_df)

```

### 6.3 ROC-AUC Curves

### 6.5 Public Health Domain Risk Analysis & Takeaways
**Domain Interpretation:**
In drinking water quality monitoring, prediction errors carry asymmetric real-world consequences:
- **False Negative (Safe water predicted as Unsafe):** Leads to unnecessary boiling or costly chemical re-treatment. Economically inconvenient, but poses zero biological risk.
- **False Positive (Unsafe/Contaminated water predicted as Safe):** Contaminated water with dangerous pH, excessive trihalomethanes, or sulfates is certified for human drinking. This directly leads to waterborne illnesses, hospitalization, and potential death.

**Conclusion:**  
Gradient Boosting and Random Forest are our premier models, achieving superior precision and discrimination (ROC-AUC > 0.88). For field deployment, probability thresholds should be set conservatively to guarantee that water declared potable is truly safe for consumption.
